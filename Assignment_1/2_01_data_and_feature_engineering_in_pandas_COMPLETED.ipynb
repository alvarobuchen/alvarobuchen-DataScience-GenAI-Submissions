### What are Dataframes?
Pandas has a very elegant way of managing data, very much borrowed from the statistical language R, mostly based around dataframes. You can think of dataframes a bit like an Excel table with rows, columns and common operations like sum, average and so on. We can use this on top of our previous work on lists and dictionaries (etc.) more usable and malleable.

To begin with we will create a dataset - in this case a [dictionary](https://github.com/MJMortensonWarwick/IB2AD0_Data_Science_GenerativeAI/blob/main/1_05_dictionaries.ipynb).



```python
orders = {'o10001':{'date':'2024/01/10', 'product':'Hoodie', 'quantity':'1'},
            'o10002':{'date':'2024/01/13', 'product':'Tote bag', 'quantity':'2'},
            'o10003':{'date':'2024/01/14', 'product':'Pencil', 'quantity':'10'},
            'o10004':{'date':'2024/01/15', 'product':'T-shirt', 'quantity':'2'}
}
orders
```

We can convert this to a Dataframe with great ease

```python
import pandas as pd
import numpy as np

orders_df = pd.DataFrame(orders)
orders_df
```

We can even create such outputs from more complex dictionaries, such as a dictionary which includes a nested list:

```python
customers = {'Mark':{'name':'Mark Johnson', 'open_orders':3, 'orders':['o10001', 'o10002', 'o10004']},
             'Katy':{'name':'Katy Hoad', 'open_orders':0, 'orders':[]},
             'Angela':{'name':'Angela Lorenz', 'open_orders':1, 'orders':['o10003']},
             'Bo':{'name':'Bo Kelestyn', 'open_orders':0, 'orders':[]}
}
customers
```

```python
customers_df = pd.DataFrame(customers)
customers_df
```

Pandas dataframes can also be built from various other data structures and sources (including Excel files, CSVs, text files, databases and many more). For example, we can make our dataframe from a set of [lists](https://github.com/MJMortensonWarwick/IB2AD0_Data_Science_GenerativeAI/blob/main/1_04_lists.ipynb):

```python
a = [1, 2, 3, 4]
b = ["a", "b", "c", "d"]
c = [True, False, True, False]

listdf = pd.DataFrame([a, b, c])
listdf
```

### Method 1: list of lists -> each list becomes a ROW

```python
cities = ["Barcelona", "London", "Coventry", "Madrid"]
population_m = [1.6, 8.9, 0.35, 3.3]
capital = [False, True, False, True]

cities_df = pd.DataFrame([cities, population_m, capital])
cities_df
```

### Method 2: dict of lists -> each list becomes a COLUMN (one row per city)

```python
cities = ["Barcelona", "London", "Coventry", "Madrid"]
population_m = [1.6, 8.9, 0.35, 3.3]
capital = [False, True, False, True]

cities_df2 = pd.DataFrame({"city": cities, "population_m": population_m, "capital": capital})
cities_df2
```

### Test: list of lists with a missing value


```python
cities = ["Barcelona", "London", "Coventry", "Madrid"]
population_m = [1.6, 8.9, 0.35, 3.3]
capital = [False, True, False]   # Madrid's value is missing

cities_missing = pd.DataFrame([cities, population_m, capital])
cities_missing
```

### Test: dict of lists with a missing value

```python
cities = ["Barcelona", "London", "Coventry", "Madrid"]
population_m = [1.6, 8.9, 0.35, 3.3]
capital = [False, True, False]   # Madrid's value is missing

try:
    cities_missing2 = pd.DataFrame({"city": cities, "population_m": population_m, "capital": capital})
    display(cities_missing2)
except ValueError as e:
    print("ValueError:", e)   # -> All arrays must be of the same length
```

### Fixes for the problems above

```python
# Fix 1: swap rows and columns (.T) so each city is a row instead of a column
display(cities_df.T)

# Fix 2: mark the missing value explicitly with np.nan so all lists have the same length
capital = [False, True, False, np.nan]
display(pd.DataFrame({"city": cities, "population_m": population_m, "capital": capital}))
```

### Test: nested dict (labelled data) with a missing value

```python
cities_dict = {
    "Barcelona": {"population_m": 1.6, "capital": False},
    "London": {"population_m": 8.9, "capital": True},
    "Coventry": {"population_m": 0.35, "capital": False},
    "Madrid": {"population_m": 3.3}   # no capital value
}
cities_dict_df = pd.DataFrame(cities_dict)
display(cities_dict_df)

# Fix 3: orient="index" gives one row per city
display(pd.DataFrame.from_dict(cities_dict, orient="index"))
```

## Findings: missing values

**Expectation:** I expected pandas to either fill the gap with an empty value

**List of lists:** This ran without an error, and the empty spot was filled with None. Pandas treats each list as a row, so it just added a blank at the end of the short row. The downside is that the cities end up as columns instead of rows, and each column holds a mix of text, numbers and True/False values.

**Dict of lists:** This raised "ValueError: All arrays must be of the same length". Each list becomes a column, and pandas cannot tell which city is missing a value, so it shows an error instead of risking that the data is misaligned

**Nested dict:** This worked, and Madrid's "capital" shows "NaN". Because the data is labelled, pandas knows exactly which value is missing.

**Conclusion:** The result partly matches my expectation. Pandas only fills a gap when it can tell where the gap is, else an error will show. Labelled data (dicts) is safer, and missing values should be marked explicitly with "np.nan" rather than left out.
